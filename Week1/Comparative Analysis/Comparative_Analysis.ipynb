```python
# Importing the necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Loading the dataset
df_comp = pd.read_excel("Comparative_Analysis.xlsx")

print(df_comp.columns.tolist(), "\n")
print(df_comp.head())
```

```python
# Structural Summary
print(df_comp.info())
```

```python
# Statistical Summary
print(df_comp.describe())
```

```python
# Frequency count of each Provider Type
print(df_comp["Provider Type"].value_counts())
```

### Chart 1 - Number of Providers per Service

```python
provider_count = df_comp.groupby("Service")["Company or Provider Name"].count().sort_values(ascending=False)

plt.figure(figsize=(12, 6))
provider_count.plot(kind="bar", color="steelblue", edgecolor="white")
plt.title("Number of Providers per Service")
plt.xlabel("Service")
plt.ylabel("Provider Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("chart1.png", dpi=150, bbox_inches="tight")
plt.show()
```

### Chart 2 - Provider Type Distribution

```python
plt.figure(figsize=(12, 7))
df_comp["Provider Type"].value_counts().plot(kind="bar", color=["#3498db", "#2ecc71", "#e74c3c"] , edgecolor="white")
plt.title("Provider Type Distribution")
plt.xlabel("Provider Type")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("chart2.png", dpi=150, bbox_inches="tight")
plt.show()
```

### Chart 3 - Provider Type Breakdown per Service

```python
cross = pd.crosstab(df_comp["Service"], df_comp["Provider Type"])

cross.plot(kind="bar", figsize=(14, 6), colormap="Set2", edgecolor="white")
plt.title("Provider Type Breakdown per Service")
plt.xlabel("Service")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Provider Type")
plt.tight_layout()
plt.savefig("chart3.png", dpi=150, bbox_inches="tight")
plt.show()
```
