```python
# Importong necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel('ILO_Nepal_2025_Dataset.xlsx', sheet_name='2025 in Numbers')
df
```

### Chart 1 - Employment Service Centre Outcomes

```python
esc = df[df['Category'] == 'Employment Service Centres']
plt.figure(figsize=(7, 4))
sns.barplot(data=esc, x='Value', y='Metric', hue='Metric', legend=False, palette='Blues_d')
plt.title('Employment Service Centre Outcomes')
plt.xlabel('Count')
plt.ylabel('')
plt.tight_layout()
plt.savefig('chart1.png', dpi=150)
plt.show()
```

### Chart 2 - Women's Share Where Reported

```python
gender = df.dropna(subset=['Women_Percent'])
plt.figure(figsize=(7, 4))
sns.barplot(data=gender, x='Women_Percent', y='Metric', hue='Metric', legend=False, palette='mako')
plt.title("Women's Share of Participation (%)")
plt.xlabel('Women (%)')
plt.ylabel('')
plt.xlim(0, 100)
plt.tight_layout()
plt.savefig('chart2.png', dpi=150)
plt.show()
```

### Chart 3 - Child Labour Free Zone Engagement vs Declaration

```python
clfz = df[df['Category'] == 'Child Labour']
plt.figure(figsize=(7, 4))
sns.barplot(data=clfz, x='Value', y='Metric', hue='Metric', legend=False, palette='crest')
plt.title('Child Labour Free Zone: Engagement vs Declaration')
plt.xlabel('Number of Local Governments')
plt.ylabel('')
plt.tight_layout()
plt.savefig('chart3.png', dpi=150)
plt.show()
```

### Chart 4 - Values by Category (All Metrics, Log Scale)

```python
plt.figure(figsize=(8, 6))
sns.barplot(data=df, x='Value', y='Metric', hue='Category', dodge=False, palette='viridis')
plt.xscale('log')
plt.title('All Reported Metrics by Category (Log Scale)')
plt.xlabel('Value (log scale)')
plt.ylabel('')
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', title='Category')
plt.tight_layout()
plt.savefig('chart4.png', dpi=150)
plt.show()
```
